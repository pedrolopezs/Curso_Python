# Laboratorio 6 · Módulo importable frente a script ejecutable

Tema 9 · Módulos

Este laboratorio compara dos ejecuciones:

- Ejecutar directamente `diagnostico.py`.
- Ejecutar `05_main_diagnostico.py`, que importa `diagnostico.py`.

El objetivo es comprobar el comportamiento de `if __name__ == "__main__"`.

```python
from pathlib import Path

# Localiza el directorio src tanto si el notebook se abre desde Tema09/notebooks
# como si VS Code usa otro directorio de trabajo cercano.
candidatos = [
    Path.cwd().parent / "src",
    Path.cwd() / "src",
    Path.home() / "Curso_Python" / "Tema09" / "src",
]

SRC = None
for candidato in candidatos:
    if candidato.exists() and (candidato / "validaciones.py").exists():
        SRC = candidato.resolve()
        break

if SRC is None:
    raise FileNotFoundError("No se encontró el directorio Tema09/src. Abre el notebook desde Tema09/notebooks o ajusta la ruta.")

print("Directorio src:", SRC)
```

```python
def mostrar_fichero(nombre: str) -> None:
    """Muestra el contenido de un fichero del directorio src."""
    ruta = SRC / nombre
    print(f"--- {ruta.relative_to(SRC)} ---")
    print(ruta.read_text(encoding="utf-8"))
```

## 1. Revisar el módulo `diagnostico.py`

```python
mostrar_fichero("diagnostico.py")
```

## 2. Ejecutar directamente `diagnostico.py`

```python
!cd "{SRC}" && python diagnostico.py
```

Al ejecutarlo directamente, el bloque `if __name__ == "__main__"` sí se ejecuta.

## 3. Revisar el script que importa `diagnostico.py`

```python
mostrar_fichero("05_main_diagnostico.py")
```

## 4. Ejecutar el script principal

```python
!cd "{SRC}" && python 05_main_diagnostico.py
```

Al importar `diagnostico.py`, sus funciones quedan disponibles, pero el bloque `__main__` no se ejecuta automáticamente.
