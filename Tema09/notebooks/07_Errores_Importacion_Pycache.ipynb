# Laboratorio 7 · Errores de importación y __pycache__

Tema 9 · Módulos

Este laboratorio revisa dos elementos prácticos:

- Errores frecuentes al importar módulos o recursos.
- Aparición del directorio `__pycache__` después de importar módulos.

```python
from pathlib import Path
import os

# Localiza el directorio src tanto si el notebook se abre desde Tema09/notebooks
# como si VS Code usa otro directorio de trabajo cercano.
candidatos = [
    Path.cwd().parent / "src",
    Path.cwd() / "src",
    Path.home() / "Curso_Python" / "Tema09" / "src",
]

SRC = None
for candidato in candidatos:
    if candidato.exists() and (candidato / "validaciones.py").exists():
        SRC = candidato.resolve()
        break

if SRC is None:
    raise FileNotFoundError("No se encontró el directorio Tema09/src. Abre el notebook desde Tema09/notebooks o ajusta la ruta.")

os.environ["TEMA09_SRC"] = str(SRC)
print("Directorio src:", SRC)
```

```python
def mostrar_fichero(nombre: str) -> None:
    """Muestra el contenido de un fichero del directorio src."""
    ruta = SRC / nombre
    print(f"--- {ruta.relative_to(SRC)} ---")
    print(ruta.read_text(encoding="utf-8"))
```

## 1. Revisar el script de errores controlados

```python
mostrar_fichero("06_errores_importacion.py")
```

## 2. Ejecutar el script

```python
%%bash
cd "$TEMA09_SRC"
python - <<'PY'
fichero_script = "06_errores_importacion.py"

with open(fichero_script, encoding="utf-8") as fichero:
    codigo = fichero.read()

namespace = {"__name__": "__main__", "__file__": fichero_script}
exec(compile(codigo, fichero_script, "exec"), namespace)
PY

```

## 3. Revisar el directorio `__pycache__`

Después de ejecutar scripts que importan módulos, Python puede generar bytecode en `__pycache__`. Este directorio no debe modificarse manualmente ni versionarse en Git.

```python
%%bash
cd "$TEMA09_SRC"
find __pycache__ -maxdepth 1 -type f -print 2>/dev/null || true
```

## 4. Diagnóstico recomendado

Si un import falla, revisar:

- nombre exacto del fichero `.py`;
- directorio desde el que se ejecuta el script;
- si el recurso importado existe dentro del módulo;
- nombres de fichero que puedan colisionar con módulos conocidos;
- importaciones circulares.
