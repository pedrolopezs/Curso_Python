# Laboratorio 8 · Ejercicio integrado con paquete propio

Tema 9 · Módulos

Este laboratorio usa un paquete propio llamado `soporte_ti`. El paquete agrupa módulos relacionados:

- `red.py`: normalización y validación básica de servicios de red.
- `reportes.py`: generación de salidas legibles.
- `__init__.py`: marca el directorio como paquete y define metadatos básicos.

```python
from pathlib import Path
import os

# Localiza el directorio src tanto si el notebook se abre desde Tema09/notebooks
# como si VS Code usa otro directorio de trabajo cercano.
candidatos = [
    Path.cwd().parent / "src",
    Path.cwd() / "src",
    Path.home() / "Curso_Python" / "Tema09" / "src",
]

SRC = None
for candidato in candidatos:
    if candidato.exists() and (candidato / "validaciones.py").exists():
        SRC = candidato.resolve()
        break

if SRC is None:
    raise FileNotFoundError("No se encontró el directorio Tema09/src. Abre el notebook desde Tema09/notebooks o ajusta la ruta.")

os.environ["TEMA09_SRC"] = str(SRC)
print("Directorio src:", SRC)
```

```python
def mostrar_fichero_relativo(nombre: str) -> None:
    """Muestra el contenido de un fichero relativo al directorio src."""
    ruta = SRC / nombre
    print(f"--- {nombre} ---")
    print(ruta.read_text(encoding="utf-8"))
```
