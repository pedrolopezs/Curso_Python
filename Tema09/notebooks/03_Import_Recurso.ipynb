# Laboratorio 3 · Importar recursos concretos

Tema 9 · Módulos

Este laboratorio ejecuta `02_import_recurso.py`. El script importa solo una función concreta desde el módulo `validaciones`.

```python
from pathlib import Path
import os

# Localiza el directorio src tanto si el notebook se abre desde Tema09/notebooks
# como si VS Code usa otro directorio de trabajo cercano.
candidatos = [
    Path.cwd().parent / "src",
    Path.cwd() / "src",
    Path.home() / "Curso_Python" / "Tema09" / "src",
]

SRC = None
for candidato in candidatos:
    if candidato.exists() and (candidato / "validaciones.py").exists():
        SRC = candidato.resolve()
        break

if SRC is None:
    raise FileNotFoundError("No se encontró el directorio Tema09/src. Abre el notebook desde Tema09/notebooks o ajusta la ruta.")

os.environ["TEMA09_SRC"] = str(SRC)
print("Directorio src:", SRC)
```

```python
def mostrar_fichero(nombre: str) -> None:
    """Muestra el contenido de un fichero del directorio src."""
    ruta = SRC / nombre
    print(f"--- {ruta.relative_to(SRC)} ---")
    print(ruta.read_text(encoding="utf-8"))
```

## 1. Revisar el script

```python
mostrar_fichero("02_import_recurso.py")
```

## 2. Ejecutar el script

```python
%%bash
cd "$TEMA09_SRC"
python - <<'PY'
fichero_script = "02_import_recurso.py"

with open(fichero_script, encoding="utf-8") as fichero:
    codigo = fichero.read()

namespace = {"__name__": "__main__", "__file__": fichero_script}
exec(compile(codigo, fichero_script, "exec"), namespace)
PY

```

## 3. Interpretación

`from validaciones import puerto_valido` solo deja disponible el nombre `puerto_valido`. Las demás funciones del módulo no quedan disponibles con su nombre directo.
