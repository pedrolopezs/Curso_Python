# Laboratorio 1 · Crear un módulo propio

Tema 9 · Módulos

En este laboratorio se revisa el módulo `validaciones.py`.

Este fichero no se ejecuta como programa principal. Su objetivo es agrupar funciones reutilizables que otros scripts podrán importar.

```python
from pathlib import Path
import os

# Localiza el directorio src tanto si el notebook se abre desde Tema09/notebooks
# como si VS Code usa otro directorio de trabajo cercano.
candidatos = [
    Path.cwd().parent / "src",
    Path.cwd() / "src",
    Path.home() / "Curso_Python" / "Tema09" / "src",
]

SRC = None
for candidato in candidatos:
    if candidato.exists() and (candidato / "validaciones.py").exists():
        SRC = candidato.resolve()
        break

if SRC is None:
    raise FileNotFoundError("No se encontró el directorio Tema09/src. Abre el notebook desde Tema09/notebooks o ajusta la ruta.")

os.environ["TEMA09_SRC"] = str(SRC)
print("Directorio src:", SRC)
```

```python
def mostrar_fichero(nombre: str) -> None:
    """Muestra el contenido de un fichero del directorio src."""
    ruta = SRC / nombre
    print(f"--- {ruta.relative_to(SRC)} ---")
    print(ruta.read_text(encoding="utf-8"))
```

## 1. Revisar el módulo `validaciones.py`

Observa que el fichero contiene funciones, docstrings y anotaciones de tipo, pero no contiene un flujo principal de ejecución.

```python
mostrar_fichero("validaciones.py")
```

## 2. Probar el módulo desde una ejecución externa

En lugar de importar el módulo directamente dentro del kernel del notebook, ejecutamos Python desde el directorio `src`. Esto reproduce el comportamiento de una terminal real.

```python
%%bash
cd "$TEMA09_SRC"
python - <<'PY'
import validaciones

print(validaciones.puerto_valido(443))
print(validaciones.normalizar_servicio(" SSH "))
print(validaciones.estado_valido(" ok "))
PY
```
